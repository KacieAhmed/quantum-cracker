# Grover's algorithm on toy seed-phrase keyspaces

Educational demo. Properly generated seed phrases (2^132 candidates) are not threatened by Grover's algorithm — see the research note *Grover's Algorithm — Mechanics, Oracle Cost, and Honest Scaling* (`art_MRbA2s2H`). This notebook shows the mechanics on toy spaces and the scaling gap.

Figures are produced headlessly by `qcracker_grover.charts`; this notebook renders them plus summary JSON from a live run.

```python
import json
from pathlib import Path

from IPython.display import Image, display

from qcracker_grover.charts import make_all
from qcracker_grover.engine import run

FIGURES = Path("figures")
paths = make_all(FIGURES, n_bits=8)
print(json.dumps([str(p) for p in paths], indent=2))
```
